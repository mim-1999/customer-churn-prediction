```python
import pandas as pd

url = 'https://raw.githubusercontent.com/mim-1999/customer-churn-prediction/main/data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv'
df = pd.read_csv(url)
```

```python
df.info()
```

```python
df.head()
```

```python
df.nunique()
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df['TotalCharges'] = df['TotalCharges'].fillna(0)
```

```python
df.describe()
```

```python
df['TotalCharges'].dtype
```

```python
df = df.drop('customerID', axis=1)
```

```python
df.nunique().sort_values()
```

```python
df['OnlineSecurity'].unique()
```

```python
df['MultipleLines'].unique()
```

```python
cols_to_simplify = ['OnlineSecurity', 'TechSupport', 'OnlineBackup',
                     'DeviceProtection', 'StreamingTV', 'StreamingMovies']
for col in cols_to_simplify:
    df[col] = df[col].replace('No internet service', 'No')
```

```python
df['MultipleLines'] = df['MultipleLines'].replace('No phone service', 'No')
```

```python
df['MultipleLines'].unique()
```

```python
df['SeniorCitizen'].unique()
```

```python
df['gender'] = df['gender'].map({'Male': 0, 'Female': 1})
df['Partner'] = df['Partner'].map({'Yes': 1, 'No': 0})
df['Dependents'] = df['Dependents'].map({'Yes': 1, 'No': 0})
df['PhoneService'] = df['PhoneService'].map({'Yes': 1, 'No': 0})
df['PaperlessBilling'] = df['PaperlessBilling'].map({'Yes': 1, 'No': 0})
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})
```

```python
df[['gender','Partner','Dependents','PhoneService','PaperlessBilling','Churn','SeniorCitizen']].dtypes
```

```python
df.nunique().sort_values()
```

```python
binary_cols = ['MultipleLines', 'OnlineBackup', 'OnlineSecurity', 'StreamingTV',
               'StreamingMovies', 'DeviceProtection', 'TechSupport']
for col in binary_cols:
    df[col] = df[col].map({'Yes': 1, 'No': 0})
```

```python
df.nunique().sort_values()
```

```python
df = pd.get_dummies(df, columns=['Contract', 'InternetService', 'PaymentMethod'], drop_first=False)
```
